## Making a GitHub PAT

The GUI interface (as of October 2025) works like this.
  * Click your profile photo.
  * Settings
  * Developer Settings (way down on the left)
  * Personal Access tokens

For this case I made a fine-grained access token that only is able to access public repos. If you deliberate mangle the `token` string below, or if you point it a private repo, this notebook won't work. 

```python
# the token is perfectly safe to publish since its scope limited, but GitHub itself doesn't understand that
# so I'm disguising it a little in this notebook
part_a = "11ABK5JYY0t1zglvX0cBkN_TxvZgDwwL4iwucC2VnQmf"
part_b = "6wsjKcwwt5dZQtV7kTIq4RKH7O4GJTS5MaIaBb"
token = part_a + part_b
token = "githu" + "b_pa" + "t_" + token
# token += "_junk" # uncomment to break the notebook and thus provve that only valid PATS will succeeed
```

```python
import github
g = github.Github(token)
```

```python
url = "https://github.com/ticdat/tts_diet" 
# url = "https://github.com/pjcpjc/nested_diet_example" # uncomment to break and thus prove only public repos work with this PAT
```

```python
user_str, repo_str = url.replace("https://github.com/", "").split("/")
```

```python
user = g.get_user(user_str) # a bad PAT will fail here
```

```python
repo = user.get_repo(repo_str) # a private repo will fail here
```

## Programmatically examine the repo

The main thing is establishing the connection, but here is a quick examination of reading from the repo. This code captures the file [here](https://github.com/ticdat/tts_diet/blob/0.0.2/tts_diet/__init__.py)

```python
sha = "0.0.2"
path = "tts_diet/__init__.py"
file_content = repo.get_contents(path, ref=sha)
blob = repo.get_git_blob(file_content.sha) # using blob here as a precaution against huge files
import base64
file_data = base64.b64decode(blob.content)
print(file_data.decode("utf-8"))
```
